```python
from lagrange import Lagrange
import numpy as np
import scipy.interpolate as interpol
import matplotlib.pyplot as plt
```

```python
lagrange = Lagrange()
```

```python
list_x = [(-5 + k) for k in range(11)]

list_y = [(lambda x: 1 / (1 + x**2))(x) for x in list_x]
```

```python
x_vals = np.linspace(-5, 5, 200)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

degrees = [1, 2, 3]

for i, deg in enumerate(degrees):
    interp_func = lagrange.interpolation_lagrange_locale(degre=deg, list_x=list_x, list_y=list_y)
    P = np.vectorize(interp_func)
    y_interp = P(x_vals)
    
    axes[i].plot(x_vals, y_interp, label=f"Interpolation deg {deg}")
    axes[i].plot(list_x, list_y, "ro", label="Points support")
    axes[i].plot(x_vals, 1/(1 + x_vals**2), "k:", label="1/(1+x²)")
    axes[i].legend()
    axes[i].set_title(f"Interpolation locale degré {deg}")

plt.tight_layout()
plt.show()
```

```python
list_x = [(-2 + k) for k in range(5)]

list_y = [(lambda x: 1 / (1 + x**2))(x) for x in list_x]

P = lagrange.interpolation_lagrange_globale(list_x=list_x, list_y=list_y)

# Grille d’abscisses pour le graphique
x = np.linspace(-2, 2, 200)

# Ordonnées calculées avec la FPI par morceaux de degré 1
y = P(x)
# Graphiques
plt.plot(x, y, label="Interpolation")
plt.plot(list_x, list_y, "ro", label="Support")
plt.plot(x, 1 / (1 + x**2), "k:", label="1/(1 + x²)")
plt.legend()
plt.title("Interpolation  globale de Lagrange de degré" + str(len(list_x) - 1))
plt.show()
```

```python
# Fonction polynomiale interpolatrice par morceaux de degr é 1
p = interpol.lagrange(list_x, list_y)

# Grille d’abscisses pour le graphique
x = np.linspace(-5, 5, 200)

# Ordonn ées calcul ées avec la FPI par morceaux de degr é 1
y = p(x)
# Graphiques
plt.plot(x, y, label="Interpolation")
plt.plot(list_x, list_y, "ro", label="Support")
plt.plot(x, 1 / (1 + x**2), "k:", label="1/(1 + x²)")
plt.legend()
plt.title("Interpolation  globale de Lagrange de degré" + str(len(list_x) - 1))
plt.show()
```
