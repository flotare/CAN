```python
from lagrange import Lagrange
import numpy as np
import scipy.interpolate as interpol
import matplotlib.pyplot as plt
```

```python
lagrange = Lagrange()
```

```python
list_x = [(-5 + k) for k in range(11)]


def f(x):
    return 1 / (1 + x**2)


list_y = [f(x) for x in list_x]
```

```python
P = lagrange.interpolation_lagrange_globale(list_x=list_x, list_y=list_y)
```

```python
# Fonction polynomiale interpolatrice par morceaux de degr é 1
# p = interpol.lagrange(list_x, list_y)

# Grille d’abscisses pour le graphique
x = np.linspace(-5, 5, 100)

# Ordonn ées calcul ées avec la FPI par morceaux de degr é 1
y = P(x)
# Graphiques
plt.plot(x, y, label="Interpolation")
plt.plot(list_x, list_y, "ro", label="Support")
plt.plot(x, 1 / (1 + x**2), "k:", label="1/(1 + x²)")
plt.legend()
plt.title("Interpolation  globale de Lagrange de degré" + str(len(list_x) - 1))
plt.show()
```
