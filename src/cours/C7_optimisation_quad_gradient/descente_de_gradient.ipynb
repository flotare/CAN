# Méthode du gradient pas constant

```python
A = np.array(
    [
        [2, -1],
        [-1, 2],
    ]
)
b = np.array(
    [
        [1],
        [0],
    ]
)

x_0 = np.array(
    [
        [3],
        [7],
    ],
    dtype=float,
)

tol = 0.01

alpha_pas = 1 / 4
```

```python
methode = DescenteGradient(A=A, b=b)
methode.solve_pas_constant(tol=tol, x_0=x_0)
```
