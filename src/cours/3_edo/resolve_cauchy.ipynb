```python
import numpy as np
import math
from matplotlib import pyplot as plt

from cauchy import Cauchy
```

# Euler Explicite

```python
t_range = (0.0, 3.0)
y0 = 0

def f(t, y):
    return np.sin(2 * t) * (1 - y)**2

cauchy = Cauchy(f=f, t_range=t_range, y0=y0)

n = 500

cauchy.euler_explicite(n=n)
```

```python
def y(t):
    return np.sin(t)**2 / (1 + np.sin(t)**2)

cauchy.erreur_calc(y=y)
```

```python
t = np.array(cauchy.list_t)

y_calc = np.array(cauchy.list_y)

y_true = y(t)

erreur = np.array(cauchy.erreur)


fig, axes = plt.subplots(1, 2, figsize=(18, 5))

axes[0].plot(t, y_calc, label="Euler Explicite")
axes[0].plot(t, y(t), "r:", label="y(t)")
axes[0].set_title(
    f"Schéma numérique Euler Explicite, y0={y0}, t0={t_range[0]}, T={t_range[1]}, N={n}"
)
axes[0].legend()

axes[1].plot(t, erreur)
axes[1].set_title(
    f"Erreur pour le schéma numérique Euler Explicite, y0={y0}, t0={t_range[0]}, T={t_range[1]}, N={n}"
)
axes[1].set_xlabel("t")
axes[1].set_ylabel("Erreur")

plt.tight_layout()
plt.show()
```

```python
list_n, list_log_erreur = cauchy.estimation_ordre(y=y, methode=cauchy.euler_explicite)

plt.plot(list_n, list_log_erreur, "+", label="Erreurs")
plt.plot(list_n, list_log_erreur, "r:", label="pente")
plt.xlabel("log(n)")
plt.ylabel("log(erreur)")
plt.legend()
plt.title(
    f"Erreur pour le schéma numérique Euler Explicite, T={t_range[1]}, ordre estimé m={math.trunc(cauchy.ordre * 100) / 100}"
)
plt.show()
```
