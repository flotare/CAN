### Vérification avec les fonctions natives

```python
B = np.array(
    [
        [1, 2, 3],
        [2, 1, 2],
        [3, 2, 1],
    ]
)
la.eig(B)
```

```python
u = np.array(
    [
        [1],
        [2],
        [3],
    ]
)
for _ in range(9):
    u = B @ u / la.norm(B @ u)

v = la.eig(B)[1][:, 0]

dot = np.dot(u.T, v)
norm = la.norm(u) * la.norm(v)

theta_rad = np.arccos(dot / norm)
theta_deg = np.degrees(theta_rad)

print(f"Angle = {float(theta_deg):.2f}°")
print(u.T@B@u)
```
