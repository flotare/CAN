```python
import scipy.integrate as itg
import numpy as np

from quadrature import Quadrature
```

```python
def f(x):
    return (x + 4) * (3 - x) * (x - 1) * (x + 1)

quad = Quadrature(f=f, a=-2, b=2, n_interval=2)
```

```python
print(quad.rectangle_gauche())
```

```python
print(quad.rectangle_droite())
```

```python
x = np.linspace(-2, 2, 3)

print(itg.trapezoid(y=f(x), dx=2))
print(quad.trapeze())
```

```python
x = np.linspace(-2, 2, 5)

print(itg.simpson(y=f(x)))
print(quad.simpson())
```
